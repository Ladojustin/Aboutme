```python
import tensorflow as tf
import numpy as np

#declaration
num_user = 1000
num_item = 500
num_sample = 10000

#for training
user_id_train = np.random.randint(0,num_user,num_sample)
user_item_train = np.random.randint(0,num_item,num_sample)
rating_train = np.random.randint(1,6,num_sample)

#for validating
user_id_val = np.random.randint(0,num_user,num_sample)
user_item_val = np.random.randint(0,num_item,num_sample)
rating_val = np.random.randint(1,6,num_sample)

#for testing
user_id_test = np.random.randint(0,num_user,num_sample)
user_item_test = np.random.randint(0,num_item,num_sample)
rating_test = np.random.randint(1,6,num_sample)

#creating class
class CreatingFilterModel(tf.keras.Model):
  def __init__(self,num_user,num_item,embedding_size):
    super(CreatingFilterModel,self).__init__()
    self.user_embedding = tf.keras.layers.Embedding(num_user,embedding_size)
    self.item_embedding = tf.keras.layers.Embedding(num_item,embedding_size)
    self.dot = tf.keras.layers.Dot(axes = 1)

  def call(self,input):
    user_id, item_id = input
    user_embedding = self.user_embedding(user_id)
    item_embedding = self.item_embedding(item_id)
    return self.dot([user_embedding,item_embedding])

embedding_size = 10
model = CreatingFilterModel(num_user,num_item, embedding_size)

model.compile(optimizer = 'adam',loss = 'mean_squared_error')

model.fit([user_id_train, user_item_train], rating_train)

loss = model.evaluate((user_id_test,user_item_test),rating_test)
print(loss)
```
